```python
import os
import sys
import subprocess
from IPython.display import Image, display

repo_url = "https://github.com/Abhishek1033ubuntu/relativistic-ecloud-tmci-mitigation-suite.git"
repo_dir = "/content/relativistic-ecloud-tmci-mitigation-suite"

# 1. Clone repository if src/ is not already in the working directory
if not os.path.exists("src"):
    if not os.path.exists(repo_dir):
        subprocess.run(["git", "clone", repo_url, repo_dir], check=True)
    os.chdir(repo_dir)

# 2. Ensure repository root is at the front of Python's module search path
repo_root = os.getcwd()
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

# 3. Run full Milestone 1, 2, and 3 validation suite
from src.run_full_validation import run_all
run_all(output_dir="results")

# 4. Display the generated validation plots directly in Colab output
for plot_file in [
    "results/milestone1_gradient_confinement.png",
    "results/milestone2_bom_ecloud_suppression.png",
    "results/milestone3_gap_gated_stabilization.png",
]:
    if os.path.exists(plot_file):
        display(Image(filename=plot_file))
```
